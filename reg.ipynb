```python
from sklearn.linear_model import LinearRegression
X=[[1],[2],[3],[4],[5]]
y=[40,50,65,75,90]
model=LinearRegression()
model.fit(X,y)
hours=float(input("Enter how many hours you studied: "))
predictmarks=model.predict([[hours]])
print(f"the marks you will get after studying {hours} hours is {predictmarks}")
```

```python
from sklearn.linear_model import LogisticRegression
X=[[6],[6.5],[7.5],[8.5],[9]]
y=[0,0,1,1,1]
model=LogisticRegression()
model.fit(X,y)
selection=float(input("Enter how much cgpa you got: "))
prediction=model.predict([[selection]])
#prediction=model.predict([[selection]])[0] from this it will convert to label like 0 or 1 
if prediction[0]==1:
    print(f"From your {selection} cgpa you are selected")
else:
    print(f"From your {selection} cgpa you are rejected")
```

```python
#kNeighbourClassifier this looks the 3 neighbours of it and predict the output
from sklearn.neighbors import KNeighborsClassifier
X=[
    [180,7],
    [200,7.5],
    [250,8],
    [300,8.5],
    [340,9],
    [360,9.5]
]
y=[0,0,0,1,1,1] #0 for apple and 1 for mango
model=KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)
weight=float(input("Enter the weight in gms: "))
size=float(input("Enter the size in cm: "))
prediction=model.predict([[weight,size]])[0]
if prediction==0:
    print("This is likely an apple")
else:
    print("This is likely an mango")
```
